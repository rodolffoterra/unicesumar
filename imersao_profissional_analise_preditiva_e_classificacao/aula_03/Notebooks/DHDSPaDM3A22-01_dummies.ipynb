# Dummies

### Introdução

Variáveis ​​fictícias ou "dummies" são freqüentemente usadas ​​em conjuntos de dados para ajudar a colocar números em objetos categóricos. Muitas vezes, quando os cientistas de dados analisam os dados, eles têm a tarefa de decompor os dados em objetos que podem ser modelados. Embora seja possível modelar objetos escritos, existem muitos objetos que só podem ser modelados com números.

Como futuros cientistas de dados, é fundamental criar números a partir de valores categóricos corretamente.

Ao treinar um modelo de aprendizado de máquina, é necessário transformar e codificar as variáveis ​​de tipo categórico. Existem dois tipos de variáveis ​​categóricas:

- [Nominal](https://www.questionpro.com/blog/nominal-ordinal-interval-ratio/). Uma variável pode ser tratada como nominal quando seus valores representam categorias que não obedecem a uma classificação intrínseca. Por exemplo, o departamento da empresa onde um funcionário trabalha. Alguns exemplos de variáveis nominais são: região, código postal ou denominação religiosa.

- [Ordinal](https://www.questionpro.com/blog/ordinal-data/). Uma variável pode ser tratada como ordinal quando seus valores representam categorias com alguma classificação intrínseca. Por exemplo, níveis de satisfação com um serviço, variando de muito insatisfeito a muito satisfeito. Exemplos de variáveis ordinais incluem escalas de atitude que representam o grau de satisfação ou confiança e pontuações de avaliação de preferência.

## Variáveis Categóricas Nominais:

```python
import pandas as pd

# carregamos o dataset 
df = pd.read_csv('../Data/Titanic.csv')
```

```python
df.head()
```

Podemos obter as variáveis `dummies` para a coluna "sexo" onde ela só pode ter dois valores possíveis: masculino ou feminino. As variáveis `dummy` obtidas representam o estado da variável categórica (presença / ausência) da mesma. Usando a função [`pandas.get_dummies()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.get_dummies.html) do pandas:

### Método 1. Pandas get_dummies

```python
dummy = pd.get_dummies(df['Sex'])
dummy.head()
```

```python
# concatenamos ambos os datasets 
df = pd.concat([df, dummy], 
               axis = 1
              )

df['Survived'] = df.Survived.apply(lambda x: 'yes' if x == 1 else 'No')
df.head()
```

```python
df.head()
```

### Método 2. `sklearn.preprocessing.OneHotEncoder`

A entrada para a função  [`Sklearn.preprocessing.OneHotEncoder()`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html) deve ser semelhante a uma matriz de inteiros ou `strings`, denotando os valores assumidos por `features` categóricas (discretas). Os recursos são codificados usando um esquema de [codificação `one-hot`](https://medium.com/@michaeldelsole/what-is-one-hot-encoding-and-how-to-do-it-f0ae272f1179) (também conhecido como `one-of-K` ou `dummy`). Isso cria uma coluna binária para cada categoria e retorna uma matriz esparsa ou matriz densa (dependendo do parâmetro esparso).

Podemos ajustar esse objeto no conjunto de treino e, em seguida, usar o mesmo objeto para transformar o conjunto de teste. Por outro lado, pd.get_dummies retorna "dataframe" com codificações baseadas nos valores do "dataframe" que passamos para ele. Isso pode ser bom para uma análise rápida, mas para um projeto que mais tarde teríamos que colocar em produção, é melhor ter o objeto OneHotEncoder para que possamos "transformar" mais tarde os novos dados que chegam ao nosso servidor.

```python
from sklearn.preprocessing import OneHotEncoder 
onehot_encoder = OneHotEncoder(handle_unknown = 'ignore')

#Ajustamos e transformamos a coluna "sex"
dummy_oneHot = onehot_encoder.fit_transform(df[['Sex']])

#Colocamos um vetor no conjunto de dados

dummy_oneHot = pd.DataFrame(dummy_oneHot.toarray(), 
                            columns = df['Sex'].unique()
                           )
dummy_oneHot.head()
```

### Variáveis `dummies` e Multicolinearidade.

A variável fictícia "trap" se manifesta diretamente a partir de uma codificação `One-hot` aplicada a variáveis categóricas. Conforme discutido anteriormente (suposições de regressão linear), o tamanho dos vetores de uma codificação `one-Hot` é igual ao número de valores exclusivos que uma coluna categórica pode assumir. Vamos imaginar a velocidade variável com $3$ níveis: 


$[1]: speed_{low} + speed_{Medium} + speed_{High} = 1$


A partir da relação acima, podemos expressar qualquer uma das três variáveis independentes em termos das outras duas, vamos pegar $ speed_{low} $ e expressá-la em termos de $ speed_ {medium} $ e $ speed_ {high} $:

$[2]: speed_{low} = 1 - speed_{Medium} - speed_{High}$

### Porque isso não é bom?

Como discutimos anteriormente, isso tem uma implicação direta no problema de regressão linear.

Podemos substituir speed_low na equação [1] por seu valor na equação [1]. Na verdade, isso significa que (pelo menos) um dos recursos com os quais estamos trabalhando é **redundante**; essa característica pode ser qualquer uma das três, já que a equação 2 pode ser escrita com qualquer uma delas. Portanto, estamos fazendo nosso modelo aprender um peso adicional que não é realmente necessário. Isso consome tempo e energia computacional. Isso também fornece uma meta de otimização que pode não ser muito razoável e também pode ser difícil de trabalhar.

### Como resolvemos isso?

Como a codificação `One-Hot` induz multicolinearidade perfeita, removemos uma das colunas da features codificada. Por exemplo, podemos escolher descartar $ speed_{medium} $ ou em nosso exemplo "masculino" embora a escolha seja completamente arbitrária.

- Quando usamos [`pandas.get_dummies()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.get_dummies.html), podemos passar um argumento adicional, `drop_first = True` para eliminar a primeira nova coluna que obtivermos após a codificação.

```python
dummy_correct = pd.get_dummies(df['Sex'], 
                               drop_first = True
                              )
dummy_correct.head()
```

Quando usamos `sklearn.preprocessing.OneHotEncoder()` e queremos descartar uma das novas colunas, passamos o argumento do construtor `drop = 'first'` da classe `OneHotEncoder`.

```python
from sklearn.preprocessing import OneHotEncoder 
onehot_encoder2 = OneHotEncoder(drop = 'first')

#Ajustamos e transformamos a coluna "sex"
dummy_oneHot_correct = onehot_encoder2.fit_transform(df[['Sex']])

#Colocamos um vetor no conjunto de dados

dummy_oneHot_correct = pd.DataFrame(dummy_oneHot_correct.toarray())
dummy_oneHot_correct.head()
```
